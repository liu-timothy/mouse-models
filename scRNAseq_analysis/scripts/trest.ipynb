```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
# BBN963sq6:
categories = ['BBN SQ6', 'BBN SQ7', 'UPPL1541']
sq6 = 2839/4837 * 100
sq7 = 3609/5441 * 100 
uppl = 5816/10165 * 100

percentages = [sq6, sq7, uppl]
```

```python
inverse_percentages = [100 - p for p in percentages]

# 2. Initialize the Plot
fig, ax = plt.subplots(figsize=(8, 3)) # Slightly wider layout for horizontal bars

# 3. Create Stacked Bars (Changed to barh)
# Left segment (Calculated Percentage)
bars1 = ax.barh(categories, percentages, label='Leukocytes', color='#3182bd', height=0.5)

# Right segment (100 - Percentage) stacked to the right
bars2 = ax.barh(categories, inverse_percentages, left=percentages, label='Non-Leukocytes', color='#e0e0e0', height=0.5)

# 4. Add Text Labels Centered inside each segment
ax.bar_label(bars1, fmt='%.1f%%', label_type='center', color='white', weight='bold', fontsize=10)
ax.bar_label(bars2, fmt='%.1f%%', label_type='center', color='black', weight='bold', fontsize=10)

# 5. Chart Styling
ax.set_xlabel('Percentage of Total Filtered Cells (%)', fontsize=11, weight='bold') # Changed to xlabel
ax.set_title('scRNA-Seq Leukocyte Populations', fontsize=12, weight='bold', pad=15)
ax.set_xlim(0, 100) # Changed to xlim
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))

# Optimize space and render
plt.tight_layout()
plt.savefig('population.png', dpi=400)

plt.show()
```
